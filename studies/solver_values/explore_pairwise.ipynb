```python
import pandas as pd

pw = pd.read_csv("results/pairwise.csv")
pw["pair"] = pw.solver_a + "-" + pw.solver_b
```

```python

print(f"rows: {len(pw)} \nper pair: {pw.pair.value_counts().to_dict()}")
print(f"\nboth pairwise rate:\n{pw.groupby(["generation_method", "pair"]).both_feasible.mean().round(3)}")
```

```python
bf = pw[pw.both_feasible]

qd2_bf = bf.groupby(["generation_method", "pair"]).q_diff_2.describe()[["count", "min", "50%", "max"]]
cdr_bf = bf.groupby(["generation_method", "pair"]).cost_diff_rel.describe()[["50%", "max"]]

print(f"q_diff_2: {qd2_bf}")
print(f"\ncost_diff_rel : {cdr_bf}")
```

```python
si = bf[bf.pair == "Ipopt-SNOPT"]
print(f"Ipopt-SNOPT both feasible: n={len(si)}")
print(f"\ncost_diff_rel: {si.cost_diff_rel.describe()[["50%", "max"]].round(6).to_dict()}")
print(f"\nq_diff_inf: {si.q_diff_inf.describe()[["50%", "max"]].round(4).to_dict()}")
print(f"\nresidual_diff: {si.common_residual_inf_diff.describe()[["count", "50%", "max"]].to_dict()}")

print(f"\nsame_q rate by pair: \n{pw.groupby("pair").same_q.mean().round(3)}")
```

```python
si = pw[(pw.pair == "Ipopt-SNOPT") & pw.both_feasible]
print(si.groupby("orientation_tol").common_residual_inf_diff.describe()[["count","50%","max"]])
print("\ncost_diff_rel by orientation_tol (joint, Ipopt-SNOPT):")
j = si[si.generation_method == "joint_space"]
print(j.groupby("orientation_tol").cost_diff_rel.describe()[["count","50%","max"]])
```
